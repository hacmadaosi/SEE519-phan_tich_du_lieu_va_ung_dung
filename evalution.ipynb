```python
import pandas as pd
df = pd.read_csv("results.csv")
df.sort_values(by="accuracy", ascending=False).head(10)
```

```python
pivot = df.pivot_table(
    index=["selector"],
    columns=["model"],
    values="accuracy"
)
print(pivot)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(pivot, annot=True)
plt.show()
```
