```python
from data_preprocessor import DataPreprocessor
from feature_selection.chi2_selector import Chi2Selector
from feature_selection.anova_selector import ANOVASelector
from feature_selection.mutual_info_selector import MutualInfoSelector
from feature_selection.rf_importance_selector import RFImportanceSelector
from feature_selection.rfe_selector import RFESelector

dp = DataPreprocessor()
datasets = dp.fit_split()

selectors = {
    "Chi2": Chi2Selector(k=10),
    "ANOVA": ANOVASelector(k=10),
    "MutualInfo": MutualInfoSelector(k=10),
    "RFImportance": RFImportanceSelector(k=10),
    "RFE": RFESelector(k=10),
}
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

models = {
    "LR": LogisticRegression(max_iter=1000),
    "KNN": KNeighborsClassifier(),
    "SVM": SVC(),
    "RF": RandomForestClassifier(random_state=42),
    "XGB": XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')
}
```

```python
from sklearn.metrics import accuracy_score
import pandas as pd

results = []

for dataset_name, data in datasets.items():
    data = datasets["all"]

    X_train = data["X_train"]
    X_test = data["X_test"]
    y_train = data["y_train"]
    y_test = data["y_test"]

    for selector_name, selector in selectors.items():
        # Fit selector
        X_train_selected = selector.fit_transform(X_train, y_train)
        X_test_selected = selector.transform(X_test)

        for model_name, model in models.items():
            # Train
            model.fit(X_train_selected, y_train)

            # Predict
            y_pred = model.predict(X_test_selected)

            # Evaluate
            acc = accuracy_score(y_test, y_pred)

            results.append(
                {
                    "dataset": dataset_name,
                    "selector": selector_name,
                    "model": model_name,
                    "accuracy": acc,
                    "features": selector.selected_features,
                }
            )

            print(f"{dataset_name} | {selector_name} | {model_name} = {acc:.4f}")
```

```python
df_results = pd.DataFrame(results)
df_results.to_csv("results.csv", index=False)

df_results.head()
```
